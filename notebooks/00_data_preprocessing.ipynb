```python
# 7개월 CSV를 월별 Parquet 파일로 변환
from pathlib import Path
import duckdb

raw_path = Path("../data/raw")
processed_path = Path("../data/processed")

processed_path.mkdir(exist_ok=True)

csv_files = sorted(raw_path.glob("*.csv"))

for csv_file in csv_files:
    parquet_file = processed_path / f"{csv_file.stem}.parquet"

    print(f"변환 중: {csv_file.name}")

    duckdb.sql(f"""
        COPY (
            SELECT *
            FROM read_csv_auto(
                '{csv_file.as_posix()}',
                header=True
            )
        )
        TO '{parquet_file.as_posix()}'
        (FORMAT PARQUET);
    """)

    print(f"완료: {parquet_file.name}")
```
